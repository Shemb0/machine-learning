#  E-commerce behavior




Overview: This dataset provides a comprehensive view of customer behavior within an e-commerce platform. Each entry in the dataset corresponds to a unique customer, offering a detailed breakdown of their interactions and transactions. The information is crafted to facilitate a nuanced analysis of customer preferences, engagement patterns, and satisfaction levels, aiding businesses in making data-driven decisions to enhance the customer experience.

Columns:

-Customer ID: Type: Numeric Description: A unique identifier assigned to each customer, ensuring distinction across the dataset.

-Gender: Type: Categorical (Male, Female) Description: Specifies the gender of the customer, allowing for gender-based analytics. Age:

-Type: Numeric Description: Represents the age of the customer, enabling age-group-specific insights. City:

-Type: Categorical (City names) Description: Indicates the city of residence for each customer, providing geographic insights.

-Membership Type: Type: Categorical (Gold, Silver, Bronze) Description: Identifies the type of membership held by the customer, influencing perks and benefits.

-Total Spend: Type: Numeric Description: Records the total monetary expenditure by the customer on the e-commerce platform.

-Items Purchased: Type: Numeric Description: Quantifies the total number of items purchased by the customer.

-Average Rating: Type: Numeric (0 to 5, with decimals) Description: Represents the average rating given by the customer for purchased items, gauging satisfaction.

-Discount Applied: Type: Boolean (True, False) Description: Indicates whether a discount was applied to the customer's purchase, influencing buying behavior.

-Days Since Last Purchase: Type: Numeric Description: Reflects the number of days elapsed since the customer's most recent purchase, aiding in retention analysis.

-Satisfaction Level: Type: Categorical (Satisfied, Neutral, Unsatisfied) Description: Captures the overall satisfaction level of the customer, providing a subjective measure of their experience.


Note: This dataset is synthetically generated for illustrative purposes, and any resemblance to real individuals or scenarios is coincidental.

#After cleaning our dataframe, we need to decide which model is best for predicting our target variable.

```python
# Modelos
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR

# Preprocesamiento
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder

# Para combinar transformaciones por columnas
from sklearn.compose import ColumnTransformer

# Para encadenar preprocesamiento + modelo
from sklearn.pipeline import Pipeline

# Para búsqueda de hiperparámetros
from sklearn.model_selection import GridSearchCV

# Para dividir datos si lo necesitás
from sklearn.model_selection import train_test_split

from sklearn.metrics import mean_squared_error

```

Taking the data into account, I decided to choose two models to use in this case.

Firstly, I considered that the dataset does not have as much data as I would have liked. For that reason, I chose SVR as a ‘linear regression–type model’, because this kind of model can fit the data better than a standard linear regression. The other model I selected was a Random Forest Regressor, since it may perform better for this target variable.


If we run the code, we can see a comparison between the model’s result and one customer from our dataset. We can also appreciate that the RMSE is very close to the original value.

```python
x = df.drop(["total spend","customer id","city","average rating","days since last purchase","satisfaction level"],axis=1)
y = df["total spend"]

x_train,x_test, y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)


numeric_features=["age","items purchased"]
onehot_features= ["gender","discount applied"]
ordinal_features=["membership type"]
ordinal_order=["Bronze","Silver","Gold"]

numeric_transformer = StandardScaler()
onehot_transformer = OneHotEncoder(handle_unknown="ignore")
ordinal_transformer = Pipeline(steps=[
    ("ordinal",OrdinalEncoder(categories=[ordinal_order])),
    ("scaler",StandardScaler()),
])

preprocessor = ColumnTransformer(
    transformers=[
      ("num",numeric_transformer,numeric_features),
      ("onehot",onehot_transformer,onehot_features),
      ("ordinal", ordinal_transformer,ordinal_features),
    ]

)

pipeline = Pipeline(steps=[
                    ("preprocessor",preprocessor),
                    ("model",SVR()),
                    ])

param_grid = [
    {
    "model": [RandomForestRegressor()],
    "model__n_estimators": [100,200],
    "model__max_depth": [None,10,20]
    },
    {

     "model" : [SVR()],
     "model__kernel":["rbf","linear"],
     "model__C": [0.1, 1, 10],
     "model__gamma":["scale","auto"]
    }
]

grid = GridSearchCV(
    estimator= pipeline,
    param_grid = param_grid,
    cv= 5,
    scoring="neg_mean_squared_error",
    n_jobs=1
)

grid.fit(x_train, y_train)

print("Mejor modelo:", grid.best_estimator_)
print("Mejores hiperparámetros:", grid.best_params_)
print("Mejor score:", grid.best_score_)


# best_model = grid.best_estimator_
# prediccion = best_model.predict(nuevo_cliente)
# print("Predicción de gasto del cliente:", prediccion[0])




pred_test = grid.predict(x_test.iloc[[65]])
print(pred_test)
mse = mean_squared_error(y_test.iloc[[65]], pred_test)
rmse = mse ** 0.5

print("MSE:", mse)
print("RMSE:", rmse)




```
